本次初查读取了 Online Retail II 的两个工作表，查看各表的字段结构和前 3 行数据，并分别统计 `Customer ID` 字段的缺失明细行数，以及缺失行数占对应工作表总明细行数的比例。

这里的统计对象是明细行，而非独立客户。同一客户可能对应多条交易明细，缺失编号的记录又无法直接通过该字段识别客户，因此不能把缺失明细行数解释为缺失客户人数。

不应将缺失的 `Customer ID` 统一填为 `0` 后，再把 `0` 当作真实客户编号参与统计。这样会把身份未知、可能属于不同客户的记录合并到同一个编号下，可能导致客户数、购买频次和复购指标失真。

当前结果揭示了客户识别和客户级分析的限制。后续复购分析需要明确可识别客户的统计范围，并披露数据覆盖情况；仅凭客户编号缺失，不能认定客户已经流失、确定缺失原因或估算实际经济损失。
